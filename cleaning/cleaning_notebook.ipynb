### Making file#3 (company)

```python
df3_org = pd.read_csv(os.path.join(parent_dir, "raw_datasets", "companies.csv"))
df4 = pd.read_csv(os.path.join(parent_dir, "raw_datasets", "company_industries.csv"))
df5 = pd.read_csv(os.path.join(parent_dir, "raw_datasets", "company_specialities.csv"))
df6 = pd.read_csv(os.path.join(parent_dir, "raw_datasets", "employee_counts.csv"))
```

```python
df3 = df3_org[['company_id', 'company_name', 'city', 'state', 'country', 'url']]
```

```python
df4.info(), df5.info(), df6.info()
```

```python
df4 = df4.drop_duplicates(subset='company_id', keep='first')
df5 = df5.drop_duplicates(subset='company_id', keep='first')
df6 = df6.drop_duplicates(subset='company_id', keep='first')

merged_df1_beta = pd.merge(df4, df5, on='company_id', how='inner').merge(df6, on='company_id', how='inner')
merged_df1 = pd.merge(df3_org[['company_id', 'company_size']], merged_df1_beta, on='company_id', how='inner')

merged_df1.info()
```

```python
df3.to_csv(os.path.join(parent_dir, "cleaned_datasets", "3_companies.csv"), index=False)
merged_df1.to_csv(os.path.join(parent_dir, "cleaned_datasets", "4_company_details.csv"), index=False)
```
